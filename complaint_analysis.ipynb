```python
import csv
from collections import Counter

counts = Counter()

with open("result.csv", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
        counts[row["complaint type"]] += int(row["count"])

print(counts.most_common(10))
```

```python
import csv
from collections import Counter

monthly_counts = Counter()

with open("311_2024.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    for row in reader:
        if row["Complaint Type"] == "Illegal Parking":
            month = row["Created Date"][:2]
            monthly_counts[month] += 1

for month in ["01", "02", "06", "07"]:
    print(f"2024-{month}: {monthly_counts[month]}")
```

```python
import matplotlib.pyplot as plt

months = ["January", "February", "June", "July"]
counts = [41124, 38792, 46370, 39867]

plt.bar(months, counts)
plt.xlabel("Month")
plt.ylabel("Number of Complaints")
plt.title("Illegal Parking Complaints in NYC (2024)")
plt.show()
```
